```python
#Reshaping, view , stacking squeezing, unsqueezing, permute
x = torch.arange(1.,11.)
x, x.shape
```

```python
#Add extra dimension (Reshaping)

x_reshaped = x.reshape(1,10) #Burda 1D olan tensoru 2Dye geciriyoruz
print(x_reshaped, x.shape)
print()
x_reshaped = x.reshape(10,1)
print(x_reshaped, x.shape)
```

```python
#Reshaping

x_reshaped = x.reshape(5,2) #10luk veriyi 2li paketledim burda veri saisi tuttugu icin bunu yapabiliyorum
x_reshaped
```

```python
#View
z = x.view(1,10)

#Changing z changes x "hardcopy"
z = x.reshape(5,2)
z[:,0] = 5
x
```

```python
#Stack
print(x)

x_stacked = torch.stack([x,x,x], dim=0) #Ust uste koy
print(x_stacked)
print()
x_stacked = torch.stack([x,x,x], dim=1) #Yan yana koy
print(x_stacked)
```

```python
#Squeeze
x_reshaped = x.reshape(1,10)

print(x_reshaped)
print(x_reshaped.shape)
print()
x_squeezed = x_reshaped.squeeze()
print(x_squeezed)
print(x_squeezed.shape)
```

```python
#Unsqueeze

print(x_squeezed)
print(x_squeezed.shape)
print()
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(x_unsqueezed)
print(x_unsqueezed.shape)
print()
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(x_unsqueezed)
print(x_unsqueezed.shape)
```

```python
#Permute - rearrange the dimensions of a target tensor in a specific order

x_image = torch.rand((224,220,3)) #Height, width, color channels

#Permute the tensor as color, height, width
x_permuted = x_image.permute(2,0,1) #Ilk hangisi gelsin seklinde calisiyo
x_permuted.shape
```

```python
#Permuted view kullaniyo yani hardcopy, bi degeri degistirirsen hardcopyden de degisir

x_image[0,0,0] = 10
print(f"First element of x_image changed to {x_image[0,0,0]}")
print(f"So the x_permuted first element also changed to {x_permuted[0,0,0]}")
```

```python
# Pytorch tensors ad Numpy
import torch
import numpy as np
```

```python
array = np.arange(1.0,8.0)
#Changing numpy to tensor
tensor = torch.from_numpy(array)#.type(torch.float32) for default
array, tensor
#Note: It changes to float64 which is not the default in pytorch
```

```python
#Change in the array reflects to the tensor "Hardcopy"
array[0] = 5
array, tensor
```

```python
#Change tensor to numpy

tensor = torch.randint(0,10,(2,5))
print(tensor)

numpy_tensor = tensor.numpy()
print(numpy_tensor)
```

```python
tensor[0,0] = 10
print(tensor)
print("It also changes the numpy")
print(numpy_tensor)
```

```python
# Reproducibility - seed in randomness

RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_A = torch.rand(3,4)
print(random_tensor_A)
torch.manual_seed(RANDOM_SEED) #Her seferide ayarlaman lazim
random_tensor_B = torch.rand(3,4)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Setup device agnostic code
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```
